```python
from google.colab import userdata
from google import genai

API_KEY = userdata.get("GEMINI_API_KEY")

client = genai.Client(api_key=API_KEY)

print("Gemini connected successfully!")
```

```python
text = """
Artificial Intelligence (AI) is transforming many industries by enabling
computers to perform tasks that normally require human intelligence.
Machine Learning allows systems to learn patterns from data and make
predictions without being explicitly programmed for every situation.
Deep Learning uses multi-layered neural networks and has achieved
significant results in areas such as image recognition, speech processing,
and natural language processing. Generative AI is another important
development that can create new content such as text, images, audio, and
code. These technologies are being used in healthcare, education,
finance, transportation, and many other fields to improve efficiency
and support decision-making.
"""

print(text)
```

```python
def generate_summary(text):

    prompt = f"""
Summarize the following text in a short and simple paragraph.

Text:
{text}

Give only the summary.
"""

    interaction = client.interactions.create(
        model="gemini-3.6-flash",
        input=prompt
    )

    return interaction.output_text.strip()


summary = generate_summary(text)

print("SUMMARY:")
print(summary)
```

```python
def generate_key_points(summary):

    prompt = f"""
Extract the most important key points from the following summary.

Summary:
{summary}

Give 4 to 5 key points.
Number each point.
"""

    interaction = client.interactions.create(
        model="gemini-3.6-flash",
        input=prompt
    )

    return interaction.output_text.strip()


key_points = generate_key_points(summary)

print("KEY POINTS:")
print(key_points)
```

```python
def generate_questions(key_points):

    prompt = f"""
Create 3 questions based on the following key points.

Key Points:
{key_points}

Give only the questions.
Number each question.
"""

    interaction = client.interactions.create(
        model="gemini-3.6-flash",
        input=prompt
    )

    return interaction.output_text.strip()


questions = generate_questions(key_points)

print("GENERATED QUESTIONS:")
print(questions)
```

```python
def prompt_chaining_system(text):

    # Chain 1: Text → Summary
    summary = generate_summary(text)

    # Chain 2: Summary → Key Points
    key_points = generate_key_points(summary)

    # Chain 3: Key Points → Questions
    questions = generate_questions(key_points)

    print("========== ORIGINAL TEXT ==========")
    print(text)

    print("\n========== SUMMARY ==========")
    print(summary)

    print("\n========== KEY POINTS ==========")
    print(key_points)

    print("\n========== GENERATED QUESTIONS ==========")
    print(questions)
```

```python
prompt_chaining_system(text)
```

```python
test_text = """
Cloud computing provides users with access to computing resources such
as storage, servers, databases, and software over the internet.
Organizations can use cloud services without maintaining all the physical
hardware themselves. Cloud computing provides benefits such as
scalability, flexibility, cost efficiency, and remote accessibility.
Major cloud service models include Infrastructure as a Service,
Platform as a Service, and Software as a Service.
"""

prompt_chaining_system(test_text)
```
