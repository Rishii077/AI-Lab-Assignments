```python
from google.colab import userdata
from google import genai

API_KEY = userdata.get("GEMINI_API_KEY")

client = genai.Client(api_key=API_KEY)

print("Gemini connected successfully!")
```

```python
customer = {
    "company": "TechNova Solutions",
    "industry": "Software and Technology",
    "size": "Medium-sized company",
    "challenge": "The company wants to automate repetitive business processes using AI.",
    "goal": "Improve productivity and reduce manual work."
}

print("Customer Information:")
print(customer)
```

```python
def research_agent(customer):

    prompt = f"""
You are a Sales Research Agent.

Analyze the following customer information:

Company: {customer["company"]}
Industry: {customer["industry"]}
Company Size: {customer["size"]}
Challenge: {customer["challenge"]}
Goal: {customer["goal"]}

Identify:
1. The customer's main business challenge.
2. Their likely need.
3. How AI could potentially help them.
4. Three useful points that a sales representative should mention.

Keep the response concise.
"""

    interaction = client.interactions.create(
        model="gemini-3.6-flash",
        input=prompt
    )

    return interaction.output_text.strip()
```

```python
def email_agent(customer, research):

    prompt = f"""
You are a Sales Email Agent.

Write a professional and personalized introductory sales email.

Customer:
Company: {customer["company"]}
Industry: {customer["industry"]}

Research:
{research}

Requirements:
- Keep the email concise.
- Personalize it for the company.
- Clearly mention how AI could help with their challenge.
- Do not make unrealistic claims.
- Include a clear call to action.
- Provide a suitable subject line.
"""

    interaction = client.interactions.create(
        model="gemini-3.6-flash",
        input=prompt
    )

    return interaction.output_text.strip()
```

```python
def review_agent(email):

    prompt = f"""
You are a Sales Email Review Agent.

Review the following sales email:

{email}

Check it for:
1. Clarity
2. Personalization
3. Professional tone
4. Relevance to the customer's challenge
5. Clear call to action

Then provide an improved final version of the email.

Return only the improved email.
"""

    interaction = client.interactions.create(
        model="gemini-3.6-flash",
        input=prompt
    )

    return interaction.output_text.strip()
```

```python
def multi_agent_sdr(customer):

    research = research_agent(customer)

    email = email_agent(customer, research)

    final_email = review_agent(email)

    print("========== CUSTOMER ==========")
    print(customer)

    print("\n========== RESEARCH AGENT ==========")
    print(research)

    print("\n========== EMAIL AGENT ==========")
    print(email)

    print("\n========== REVIEW AGENT / FINAL EMAIL ==========")
    print(final_email)
```

```python
multi_agent_sdr(customer)
```
