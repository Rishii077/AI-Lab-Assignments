```python
from google.colab import userdata
from google import genai

API_KEY = userdata.get("GEMINI_API_KEY")

client = genai.Client(api_key=API_KEY)

print("Gemini connected successfully!")
```

```python
market_data = """
We are looking for potential business customers for an AI automation
solution.

The solution can help companies automate repetitive tasks, improve
employee productivity, reduce manual work, and organize business
processes using Artificial Intelligence.

Potential industries include:
- Software and Technology
- Healthcare
- Finance
- Retail
- Education

Target companies should preferably be medium-sized or large companies
that have repetitive manual processes or are interested in AI
automation.
"""

print("Market information created successfully!")
```

```python
def lead_generation_agent(market_data):

    prompt = f"""
You are a Lead Generation Agent.

Analyze the following market information:

{market_data}

Generate 5 realistic but SYNTHETIC potential business leads.

For each lead provide:

1. Company name
2. Industry
3. Company size
4. Possible business challenge
5. Potential AI requirement

Do not use real personal names or personal contact information.

Present the results clearly.
"""

    interaction = client.interactions.create(
        model="gemini-3.6-flash",
        input=prompt
    )

    return interaction.output_text.strip()
```

```python
def lead_qualification_agent(leads):

    prompt = f"""
You are a Lead Qualification Agent.

Below are potential business leads:

{leads}

Evaluate each lead based on:

1. Company size
2. Business challenge
3. Potential need for AI automation
4. Likelihood of being interested in the solution

Classify each lead as:

QUALIFIED
or
NOT QUALIFIED

For each lead provide a short reason.

Do not invent personal contact information.
"""

    interaction = client.interactions.create(
        model="gemini-3.6-flash",
        input=prompt
    )

    return interaction.output_text.strip()
```

```python
def email_agent(qualified_leads):

    prompt = f"""
You are a Sales Email Agent.

Use the following qualified leads:

{qualified_leads}

Create a short professional introductory sales email for the
qualified leads.

The email should:

- Be personalized to the company's industry and challenge.
- Explain how AI automation could potentially help.
- Avoid unrealistic claims.
- Include a suitable subject line.
- Include a clear call to action.
- Keep the email concise.

Do not include fake personal names, phone numbers, or email addresses.
"""

    interaction = client.interactions.create(
        model="gemini-3.6-flash",
        input=prompt
    )

    return interaction.output_text.strip()
```

```python
def multi_agent_sdr(market_data):

    print("========== MULTI-AGENT SDR SYSTEM ==========")

    # Agent 1: Lead Generation
    leads = lead_generation_agent(market_data)

    print("\n========== LEAD GENERATION AGENT ==========")
    print(leads)

    # Agent 2: Lead Qualification
    qualified_leads = lead_qualification_agent(leads)

    print("\n========== LEAD QUALIFICATION AGENT ==========")
    print(qualified_leads)

    # Agent 3: Email Generation
    emails = email_agent(qualified_leads)

    print("\n========== EMAIL AGENT ==========")
    print(emails)

    return emails
```

```python
final_emails = multi_agent_sdr(market_data)
```
