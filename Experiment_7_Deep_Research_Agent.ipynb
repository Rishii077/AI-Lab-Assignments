```python
from google.colab import userdata
from google import genai

API_KEY = userdata.get("GEMINI_API_KEY")

client = genai.Client(api_key=API_KEY)

print("Gemini connected successfully!")
```

```python
research_topic = """
Applications of Artificial Intelligence in Healthcare
"""

print("Research topic:")
print(research_topic)
```

```python
def planning_agent(topic):

    prompt = f"""
You are a Deep Research Planning Agent.

Research Topic:
{topic}

Create a structured research plan.

The plan should contain:

1. Introduction
2. Major applications
3. Benefits
4. Challenges and limitations
5. Future scope
6. Conclusion

Also create 3 important research questions
that should be answered while writing the content.

Keep the plan concise and well organized.
"""

    interaction = client.interactions.create(
        model="gemini-3.6-flash",
        input=prompt
    )

    return interaction.output_text.strip()
```

```python
research_plan = planning_agent(research_topic)

print("========== RESEARCH PLAN ==========")
print(research_plan)
```

```python
def content_generation_agent(topic, plan):

    prompt = f"""
You are a Research Content Generation Agent.

Research Topic:
{topic}

Research Plan:
{plan}

Using the research plan, generate a structured research article.

Requirements:

- Include an introduction.
- Explain the major applications.
- Explain important benefits.
- Discuss challenges and limitations.
- Discuss future scope.
- End with a conclusion.
- Use clear and simple academic language.
- Organize the content using headings.
- Do not invent specific statistics or research citations.
"""

    interaction = client.interactions.create(
        model="gemini-3.6-flash",
        input=prompt
    )

    return interaction.output_text.strip()
```

```python
research_topic = "Applications of Artificial Intelligence in Healthcare"
```

```python
research_plan = planning_agent(research_topic)

print("========== RESEARCH PLAN ==========")
print(research_plan)
```

```python
initial_content = content_generation_agent(
    research_topic,
    research_plan
)

print("========== INITIAL CONTENT ==========")
print(initial_content)
```

```python
def reflection_agent(topic, plan, content):

    prompt = f"""
You are a Research Reflection Agent.

Research Topic:
{topic}

Original Research Plan:
{plan}

Generated Content:
{content}

Review the generated content carefully.

Check whether:

1. All important sections from the research plan are covered.
2. The content is relevant to the research topic.
3. The explanation is clear and logically organized.
4. Important aspects are missing.
5. There are unsupported or overly specific claims.
6. The conclusion is appropriate.

Provide:

- Strengths
- Missing information
- Areas for improvement
- Specific recommendations for revision
"""

    interaction = client.interactions.create(
        model="gemini-3.6-flash",
        input=prompt
    )

    return interaction.output_text.strip()
```

```python
reflection = reflection_agent(
    research_topic,
    research_plan,
    initial_content
)

print("========== REFLECTION ==========")
print(reflection)
```

```python
def revision_agent(topic, plan, content, reflection):

    prompt = f"""
You are a Research Revision Agent.

Research Topic:
{topic}

Research Plan:
{plan}

Initial Content:
{content}

Reflection:
{reflection}

Improve the research content based on the reflection.

Requirements:

- Preserve useful information from the original content.
- Add missing important information.
- Improve organization and clarity.
- Correct weak or unsupported statements.
- Maintain simple academic language.
- Keep the original research topic.
- Include introduction, applications, benefits,
  challenges, future scope and conclusion.
"""

    interaction = client.interactions.create(
        model="gemini-3.6-flash",
        input=prompt
    )

    return interaction.output_text.strip()
```

```python
final_content = revision_agent(
    research_topic,
    research_plan,
    initial_content,
    reflection
)

print("========== FINAL RESEARCH CONTENT ==========")
print(final_content)
```
