```python
from google.colab import userdata
from google import genai

API_KEY = userdata.get("GEMINI_API_KEY")

client = genai.Client(api_key=API_KEY)

print("Gemini connected successfully!")
```

```python
benchmark_question = """
A college has 120 students.

60 students know Python.
50 students know Java.
20 students know both Python and Java.

How many students know at least one of the two programming languages?

Explain your answer.
"""

print(benchmark_question)
```

```python
def direct_prompt(question):

    prompt = f"""
Answer the following question directly.

Question:
{question}

Give a clear and concise answer.
"""

    interaction = client.interactions.create(
        model="gemini-3.6-flash",
        input=prompt
    )

    return interaction.output_text.strip()
```

```python
def structured_prompt(question):

    prompt = f"""
Solve the following question using a structured approach.

Question:
{question}

Provide:
1. Given information
2. Formula or method
3. Calculation
4. Final answer
"""

    interaction = client.interactions.create(
        model="gemini-3.6-flash",
        input=prompt
    )

    return interaction.output_text.strip()
```

```python
def reasoning_prompt(question):

    prompt = f"""
Solve the following problem carefully.

Question:
{question}

Analyze the relationships between the given values,
identify the correct mathematical method, and verify
the result before giving the final answer.

Present the important reasoning steps clearly,
followed by the final answer.
"""

    interaction = client.interactions.create(
        model="gemini-3.6-flash",
        input=prompt
    )

    return interaction.output_text.strip()
```

```python
direct_result = direct_prompt(benchmark_question)

structured_result = structured_prompt(benchmark_question)

reasoning_result = reasoning_prompt(benchmark_question)

print("========== DIRECT PROMPT ==========")
print(direct_result)

print("\n========== STRUCTURED PROMPT ==========")
print(structured_result)

print("\n========== REASONING PROMPT ==========")
print(reasoning_result)
```

```python
comparison = {
    "Direct Prompt": direct_result,
    "Structured Prompt": structured_result,
    "Reasoning Prompt": reasoning_result
}

print("Results stored successfully!")
```

```python
expected_answer = "90"

print("========== BENCHMARK EVALUATION ==========")

for strategy, result in comparison.items():

    if expected_answer in result:
        status = "Correct"
    else:
        status = "Needs Review"

    print(strategy, ":", status)
```
